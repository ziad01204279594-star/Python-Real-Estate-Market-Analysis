```python
df[['PRICE', 'BEDS', 'BATH', 'PROPERTYSQFT', 'TYPE']].isnull().sum()
```

```python
df_filtered = df[df['PRICE'] < 5_000_000].copy()
df_filtered.shape
```

```python
df_filtered['PRICE'].plot.hist(
    bins=20,
    title='Distribution of House Prices in New York'
)
plt.xlabel('Price')
plt.show()
```

```python
df_filtered.plot.scatter(
    x='BEDS',
    y='PRICE',
    title='Bedrooms vs Price'
)
plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.scatter(
    df_filtered['BATH'],
    df_filtered['PRICE'],
    c=df_filtered['BEDS'],  
    cmap='viridis',
    alpha=.8
)

plt.colorbar(label='Number of Bedrooms')
plt.title('Bathrooms vs Price (Colored by Bedrooms)')
plt.xlabel('Bathrooms')
plt.ylabel('Price')
plt.grid(True)

plt.show()
```

```python
df_filtered['TYPE'].value_counts().head(10).plot.bar(
    title='Top 10 Property Types'
)
plt.xlabel('Property Type')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.scatter(
    df_filtered['PROPERTYSQFT'],
    df_filtered['PRICE'],
    c=df_filtered['BATH'],  
    cmap='plasma',
    alpha=0.8
)

plt.colorbar(label='Bathrooms')
plt.title('Property Size vs Price (Colored by Bathrooms)')
plt.xlabel('Square Footage')
plt.ylabel('Price')
plt.grid(True)

plt.show()
```
